importing all the libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score,davies_bouldin_score
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
x = iris.data
y = iris.target
```

```python
scale = StandardScaler()
```

```python
x_scaled = scale.fit_transform(x)
```

```python
kmeans = KMeans(n_clusters=3,random_state = 42)
```

```python
y_kmeans = kmeans.fit_predict(x_scaled)
```

```python
print(kmeans.cluster_centers_)
```

```python
#metric (wcss) ->should be low
print(kmeans.inertia_)
```

```python
print("cluster labels assigned to each data point",y_kmeans)
```

```python
print("cluster labels assigned to each data point",y_kmeans)
```

```python
#metrics
```

```python
print("silhouette score",silhouette_score(x_scaled,y_kmeans))
```

```python
print("davies bouldin score",davies_bouldin_score(x_scaled,y_kmeans))
```

first ten value comparision

```python
comparision=pd.DataFrame({'actual' : y,'predicted' :y_kmeans})
print(comparision)
```

```python
import matplotlib.pyplot as plt

plt.scatter(x_scaled[:,0],x_scaled[:,1],c=y_kmeans,cmap='rainbow')
plt.scatter(kmeans.cluster_centers_[0],kmeans.cluster_centers_[1],s=200,marker = 'X',label='centroid')
plt.title("KMEANS CLUSTERING ON IRIS DATASET")
plt.xlabel("feature 1(scaled)")
plt.ylabel("feature 2(scaled)")
```
